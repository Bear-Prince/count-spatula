```python
from build123d import (
    BuildPart,
    BuildSketch,
    Mode,
    Rectangle,
    Circle,
    Plane,
    extrude,
    fillet,
    mirror,
    export_stl
)
from ocp_vscode import show

# Create a base sketch on the XY plane
with BuildPart() as part:
    with BuildSketch(Plane.XY) as sketch:
        Rectangle(50, 30)  # 50mm x 30mm rectangle
        Circle(5, mode=Mode.SUBTRACT)  # Hole in the center

    # Extrude the sketch into a 3D part
    extrude(amount=10)

    # Apply a fillet to all edges
    fillet(part.edges(), radius=2)

    # Mirror the part across the YZ plane
    mirror(about=Plane.YZ)

show(part)

# Export the part to STL
# export_stl(part.part, "example_part.stl")
```
